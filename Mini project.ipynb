```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load cleaned dataset
df = pd.read_excel("employee_attrition_cleaned_dataset.xlsx")

# Display dataset
print(df.head())
print("\nDataset Shape:", df.shape)

# -------------------------------
# 1. Attrition Distribution
# -------------------------------
plt.figure(figsize=(6, 5))
sns.countplot(data=df, x="Attrition")
plt.title("Employee Attrition Distribution")
plt.xlabel("Attrition")
plt.ylabel("Number of Employees")
plt.show()


# -------------------------------
# 2. Attrition by Department
# -------------------------------
dept_attrition = pd.crosstab(
    df["Department"],
    df["Attrition"],
    normalize="index"
) * 100

dept_attrition.plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Attrition Rate by Department")
plt.xlabel("Department")
plt.ylabel("Attrition Rate (%)")
plt.xticks(rotation=45)
plt.legend(title="Attrition")
plt.tight_layout()
plt.show()


# -------------------------------
# 3. Attrition by Job Role
# -------------------------------
role_attrition = pd.crosstab(
    df["Job Role"],
    df["Attrition"],
    normalize="index"
) * 100

role_attrition.plot(
    kind="bar",
    figsize=(9, 5)
)

plt.title("Attrition Rate by Job Role")
plt.xlabel("Job Role")
plt.ylabel("Attrition Rate (%)")
plt.xticks(rotation=45)
plt.legend(title="Attrition")
plt.tight_layout()
plt.show()


# -------------------------------
# 4. Attrition by Overtime
# -------------------------------
overtime_attrition = pd.crosstab(
    df["Overtime"],
    df["Attrition"],
    normalize="index"
) * 100

overtime_attrition.plot(
    kind="bar",
    figsize=(7, 5)
)

plt.title("Attrition Rate by Overtime")
plt.xlabel("Overtime")
plt.ylabel("Attrition Rate (%)")
plt.xticks(rotation=0)
plt.legend(title="Attrition")
plt.tight_layout()
plt.show()


# -------------------------------
# 5. Attrition by Job Satisfaction
# -------------------------------
satisfaction_attrition = pd.crosstab(
    df["Job Satisfaction"],
    df["Attrition"],
    normalize="index"
) * 100

satisfaction_attrition.plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Attrition Rate by Job Satisfaction")
plt.xlabel("Job Satisfaction")
plt.ylabel("Attrition Rate (%)")
plt.xticks(rotation=0)
plt.legend(title="Attrition")
plt.tight_layout()
plt.show()


# -------------------------------
# 6. Age vs Monthly Income
# -------------------------------
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Age",
    y="Monthly Income",
    hue="Attrition",
    s=100
)

plt.title("Age vs Monthly Income by Attrition")
plt.xlabel("Age")
plt.ylabel("Monthly Income")
plt.tight_layout()
plt.show()


# -------------------------------
# 7. Years at Company vs Attrition
# -------------------------------
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="Attrition",
    y="Years at Company"
)

plt.title("Years at Company vs Attrition")
plt.xlabel("Attrition")
plt.ylabel("Years at Company")
plt.tight_layout()
plt.show()


# -------------------------------
# 8. Work-Life Balance vs Attrition
# -------------------------------
balance_attrition = pd.crosstab(
    df["Work-Life Balance"],
    df["Attrition"],
    normalize="index"
) * 100

balance_attrition.plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Attrition Rate by Work-Life Balance")
plt.xlabel("Work-Life Balance")
plt.ylabel("Attrition Rate (%)")
plt.xticks(rotation=0)
plt.legend(title="Attrition")
plt.tight_layout()
plt.show()


# -------------------------------
# 9. Gender vs Attrition
# -------------------------------
gender_attrition = pd.crosstab(
    df["Gender"],
    df["Attrition"],
    normalize="index"
) * 100

gender_attrition.plot(
    kind="bar",
    figsize=(7, 5)
)

plt.title("Attrition Rate by Gender")
plt.xlabel("Gender")
plt.ylabel("Attrition Rate (%)")
plt.xticks(rotation=0)
plt.legend(title="Attrition")
plt.tight_layout()
plt.show()


# -------------------------------
# 10. Correlation Heatmap
# -------------------------------
numeric_cols = [
    "Age",
    "Monthly Income",
    "Job Satisfaction",
    "Work-Life Balance",
    "Years at Company",
    "Performance Rating"
]

plt.figure(figsize=(9, 6))

sns.heatmap(
    df[numeric_cols].corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap of HR Attributes")
plt.tight_layout()
plt.show()
```
